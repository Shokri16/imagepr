```python

import cv2
import math

# ------------------------------------------------
# Load the video
# ------------------------------------------------
video_path = r"\\192.168.1.10\shear all\day4\british_highway_traffic.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
    exit()

# ------------------------------------------------
# Video information
# ------------------------------------------------
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

duration = frame_count / fps if fps > 0 else 0

print("Video opened successfully!")
print("-----------------------------")
print(f"Width: {width} pixels")
print(f"Height: {height} pixels")
print(f"FPS: {fps:.2f}")
print(f"Frame count: {frame_count}")
print(f"Duration: {duration:.2f} seconds")
print("-----------------------------")


# ------------------------------------------------
# Background Subtraction
# ------------------------------------------------
background_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=50,
    detectShadows=True
)


# ------------------------------------------------
# Object tracking variables
# ------------------------------------------------
next_object_id = 0

# Dictionary:
# ID -> (center_x, center_y)
objects = {}

# Maximum distance allowed between old and new center
max_distance = 50

# Minimum object area
min_area = 500


# ------------------------------------------------
# Function to calculate distance
# ------------------------------------------------
def calculate_distance(point1, point2):
    return math.sqrt(
        (point1[0] - point2[0]) ** 2 +
        (point1[1] - point2[1]) ** 2
    )


# ------------------------------------------------
# Read video continuously
# ------------------------------------------------
frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # ------------------------------------------------
    # 1. Background Subtraction
    # ------------------------------------------------
    fg_mask = background_subtractor.apply(frame)

    # ------------------------------------------------
    # 2. Remove shadows and noise
    # ------------------------------------------------
    _, fg_mask = cv2.threshold(
        fg_mask,
        200,
        255,
        cv2.THRESH_BINARY
    )

    # ------------------------------------------------
    # 3. Morphological operations
    # ------------------------------------------------
    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    # Remove small noise
    fg_mask = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_OPEN,
        kernel
    )

    # Connect nearby regions
    fg_mask = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_CLOSE,
        kernel
    )

    # Expand objects slightly
    fg_mask = cv2.dilate(
        fg_mask,
        kernel,
        iterations=2
    )


    # ------------------------------------------------
    # 4. Find contours
    # ------------------------------------------------
    contours, _ = cv2.findContours(
        fg_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    # ------------------------------------------------
    # Store detected objects in current frame
    # ------------------------------------------------
    current_objects = []

    for contour in contours:

        # Calculate area
        area = cv2.contourArea(contour)

        # Ignore small objects/noise
        if area < min_area:
            continue

        # Bounding box
        x, y, w, h = cv2.boundingRect(contour)

        # Center of object
        center_x = x + w // 2
        center_y = y + h // 2

        current_objects.append(
            (center_x, center_y, x, y, w, h)
        )


    # ------------------------------------------------
    # 5. Object Tracking
    # ------------------------------------------------

    new_objects = {}

    # Keep track of IDs that have already been used
    used_ids = set()

    for center_x, center_y, x, y, w, h in current_objects:

        current_center = (center_x, center_y)

        best_id = None
        best_distance = max_distance

        # Compare with previous objects
        for object_id, previous_center in objects.items():

            if object_id in used_ids:
                continue

            distance = calculate_distance(
                current_center,
                previous_center
            )

            if distance < best_distance:
                best_distance = distance
                best_id = object_id

        # If no matching object was found,
        # create a new ID
        if best_id is None:

            best_id = next_object_id
            next_object_id += 1

        # Save object
        new_objects[best_id] = current_center
        used_ids.add(best_id)

        # ------------------------------------------------
        # 6. Draw Bounding Box
        # ------------------------------------------------
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Draw center point
        cv2.circle(
            frame,
            (center_x, center_y),
            5,
            (0, 0, 255),
            -1
        )

        # Display object ID
        cv2.putText(
            frame,
            f"ID: {best_id}",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )


    # Update objects
    objects = new_objects


    # ------------------------------------------------
    # 7. Display frame number
    # ------------------------------------------------
    cv2.putText(
        frame,
        f"Frame: {frame_number}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    # Number of detected objects
    cv2.putText(
        frame,
        f"Objects: {len(current_objects)}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    # ------------------------------------------------
    # 8. Display results
    # ------------------------------------------------

    cv2.imshow("Object Detection and Tracking", frame)

    cv2.imshow("Background Subtraction", fg_mask)


    print(
        f"Frame: {frame_number} | "
        f"Objects detected: {len(current_objects)}"
    )

    frame_number += 1


    # ------------------------------------------------
    # 9. ESC or q to stop
    # ------------------------------------------------
    key = cv2.waitKey(30) & 0xFF

    if key == 27 or key == ord('q'):
        break


# ------------------------------------------------
# Release everything
# ------------------------------------------------
cap.release()
cv2.destroyAllWindows()

print("Finished.")

```

```python
import cv2
import math
import numpy as np


# ============================================================
# SETTINGS
# ============================================================

video_path = r"\\192.168.1.10\shear all\day4\dhaka_traffic.mp4"

MOTION_THRESHOLD = 8

MIN_AREA_FAR = 30
MIN_AREA_NEAR = 300

MAX_DISTANCE = 100
MAX_MISSING = 8

ALPHA = 0.6


# ============================================================
# ROAD ROI
# Adjust these 4 values if needed
# ============================================================

ROAD_TOP_Y = 0.38
ROAD_TOP_LEFT = 0.28
ROAD_TOP_RIGHT = 0.72


# ============================================================
# OPEN VIDEO
# ============================================================

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
    exit()


width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

duration = frame_count / fps if fps > 0 else 0


print("Video opened successfully!")
print("-----------------------------")
print(f"Width: {width} pixels")
print(f"Height: {height} pixels")
print(f"FPS: {fps:.2f}")
print(f"Frame count: {frame_count}")
print(f"Duration: {duration:.2f} seconds")
print("-----------------------------")


# ============================================================
# MORPHOLOGY
# ============================================================

kernel = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (5, 5)
)


# ============================================================
# ROAD MASK
# ============================================================

road_mask = np.zeros(
    (height, width),
    dtype=np.uint8
)

road_points = np.array([
    [
        int(width * ROAD_TOP_LEFT),
        int(height * ROAD_TOP_Y)
    ],
    [
        int(width * ROAD_TOP_RIGHT),
        int(height * ROAD_TOP_Y)
    ],
    [
        width,
        height
    ],
    [
        0,
        height
    ]
], dtype=np.int32)

cv2.fillPoly(
    road_mask,
    [road_points],
    255
)


# ============================================================
# FEATURE REGION FOR CAMERA MOTION
# Use mostly upper background
# ============================================================

feature_mask = np.zeros(
    (height, width),
    dtype=np.uint8
)

feature_mask[
    :int(height * 0.55),
    :
] = 255


# ============================================================
# TRACKING
# ============================================================

next_object_id = 0
objects = {}


def calculate_distance(point1, point2):

    return math.sqrt(
        (point1[0] - point2[0]) ** 2 +
        (point1[1] - point2[1]) ** 2
    )


# ============================================================
# ADAPTIVE AREA
# ============================================================

def get_min_area(y, h):

    bottom = y + h

    position = bottom / height

    position = max(
        0.0,
        min(1.0, position)
    )

    area = (
        MIN_AREA_FAR +
        (
            MIN_AREA_NEAR -
            MIN_AREA_FAR
        ) *
        (position ** 2)
    )

    return int(area)


# ============================================================
# MERGE FRAGMENTED OBJECTS
# ============================================================

def boxes_should_merge(box1, box2):

    x1, y1, w1, h1 = box1
    x2, y2, w2, h2 = box2


    right1 = x1 + w1
    bottom1 = y1 + h1

    right2 = x2 + w2
    bottom2 = y2 + h2


    horizontal_overlap = (
        min(right1, right2) -
        max(x1, x2)
    )

    vertical_overlap = (
        min(bottom1, bottom2) -
        max(y1, y2)
    )


    horizontal_gap = max(
        x1 - right2,
        x2 - right1,
        0
    )

    vertical_gap = max(
        y1 - bottom2,
        y2 - bottom1,
        0
    )


    if vertical_overlap > 0.60 * min(h1, h2):

        if horizontal_gap <= 5:
            return True


    if horizontal_overlap > 0.60 * min(w1, w2):

        if vertical_gap <= 5:
            return True


    intersection_width = max(
        0,
        horizontal_overlap
    )

    intersection_height = max(
        0,
        vertical_overlap
    )


    intersection_area = (
        intersection_width *
        intersection_height
    )


    smaller_area = min(
        w1 * h1,
        w2 * h2
    )


    if smaller_area > 0:

        overlap_ratio = (
            intersection_area /
            smaller_area
        )

        if overlap_ratio > 0.50:
            return True


    return False


def merge_boxes(box1, box2):

    x1, y1, w1, h1 = box1
    x2, y2, w2, h2 = box2


    x = min(
        x1,
        x2
    )

    y = min(
        y1,
        y2
    )


    right = max(
        x1 + w1,
        x2 + w2
    )

    bottom = max(
        y1 + h1,
        y2 + h2
    )


    return (
        x,
        y,
        right - x,
        bottom - y
    )


# ============================================================
# FIRST FRAME
# ============================================================

ret, previous_frame = cap.read()

if not ret:
    print("Error: Could not read first frame.")
    cap.release()
    exit()


previous_gray = cv2.cvtColor(
    previous_frame,
    cv2.COLOR_BGR2GRAY
)

previous_gray = cv2.GaussianBlur(
    previous_gray,
    (5, 5),
    0
)


frame_number = 1


# ============================================================
# MAIN LOOP
# ============================================================

while True:

    ret, frame = cap.read()

    if not ret:
        break


    # ========================================================
    # CURRENT FRAME
    # ========================================================

    current_gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    current_gray = cv2.GaussianBlur(
        current_gray,
        (5, 5),
        0
    )


    # ========================================================
    # FIND BACKGROUND FEATURES
    # ========================================================

    previous_points = cv2.goodFeaturesToTrack(
        previous_gray,
        maxCorners=150,
        qualityLevel=0.01,
        minDistance=12,
        blockSize=7,
        mask=feature_mask
    )


    # ========================================================
    # ESTIMATE HORIZONTAL CAMERA MOVEMENT
    # ========================================================

    dx = 0.0
    dy = 0.0


    if previous_points is not None:

        current_points, status, _ = cv2.calcOpticalFlowPyrLK(
            previous_gray,
            current_gray,
            previous_points,
            None,
            winSize=(15, 15),
            maxLevel=2
        )


        if current_points is not None:

            good_previous = previous_points[
                status.ravel() == 1
            ]

            good_current = current_points[
                status.ravel() == 1
            ]


            if len(good_previous) >= 10:

                movement = (
                    good_current -
                    good_previous
                ).reshape(-1, 2)


                # Remove extreme feature movements
                movement_x = movement[:, 0]
                movement_y = movement[:, 1]


                valid_x = movement_x[
                    np.abs(movement_x) < 30
                ]

                valid_y = movement_y[
                    np.abs(movement_y) < 15
                ]


                if len(valid_x) >= 10:
                    dx = float(
                        np.median(valid_x)
                    )


                if len(valid_y) >= 10:
                    dy = float(
                        np.median(valid_y)
                    )


    # ========================================================
    # ALIGN CURRENT FRAME TO PREVIOUS FRAME
    # ========================================================

    transform = np.array([
        [1, 0, -dx],
        [0, 1, -dy]
    ], dtype=np.float32)


    aligned_current = cv2.warpAffine(
        current_gray,
        transform,
        (width, height),
        flags=cv2.INTER_LINEAR,
        borderMode=cv2.BORDER_REPLICATE
    )


    # ========================================================
    # FRAME DIFFERENCE
    # ========================================================

    difference = cv2.absdiff(
        previous_gray,
        aligned_current
    )


    # ========================================================
    # THRESHOLD
    # ========================================================

    _, mask = cv2.threshold(
        difference,
        MOTION_THRESHOLD,
        255,
        cv2.THRESH_BINARY
    )


    # ========================================================
    # ROAD ONLY
    # ========================================================

    mask = cv2.bitwise_and(
        mask,
        road_mask
    )


    # ========================================================
    # MORPHOLOGICAL CLOSING
    # ========================================================

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        kernel,
        iterations=1
    )


    # ========================================================
    # CONNECTED COMPONENTS
    # ========================================================

    number_of_labels, labels, stats, centroids = (
        cv2.connectedComponentsWithStats(
            mask,
            connectivity=8
        )
    )


    clean_mask = np.zeros_like(
        mask
    )


    for label in range(
        1,
        number_of_labels
    ):

        x = stats[
            label,
            cv2.CC_STAT_LEFT
        ]

        y = stats[
            label,
            cv2.CC_STAT_TOP
        ]

        w = stats[
            label,
            cv2.CC_STAT_WIDTH
        ]

        h = stats[
            label,
            cv2.CC_STAT_HEIGHT
        ]

        area = stats[
            label,
            cv2.CC_STAT_AREA
        ]


        min_area = get_min_area(
            y,
            h
        )


        if area < min_area:
            continue


        clean_mask[
            labels == label
        ] = 255


    # ========================================================
    # FIND CONTOURS
    # ========================================================

    contours, _ = cv2.findContours(
        clean_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    # ========================================================
    # VALID CONTOURS
    # ========================================================

    valid_contours = []


    for contour in contours:

        area = cv2.contourArea(
            contour
        )


        x, y, w, h = cv2.boundingRect(
            contour
        )


        min_area = get_min_area(
            y,
            h
        )


        if area < min_area:
            continue


        if w < 3 or h < 3:
            continue


        # Ignore objects touching the upper ROI boundary.
        # These are often building/background remnants.
        if y <= int(height * ROAD_TOP_Y) + 3:
            continue


        valid_contours.append(
            contour
        )


    # ========================================================
    # FILL CONTOUR INTERIORS
    # ========================================================

    final_mask = np.zeros_like(
        mask
    )


    cv2.drawContours(
        final_mask,
        valid_contours,
        -1,
        255,
        -1
    )


    # ========================================================
    # CREATE BOXES
    # ========================================================

    boxes = []


    for contour in valid_contours:

        x, y, w, h = cv2.boundingRect(
            contour
        )


        boxes.append(
            (
                x,
                y,
                w,
                h
            )
        )


    # ========================================================
    # MERGE FRAGMENTS
    # ========================================================

    merged = True


    while merged:

        merged = False

        new_boxes = []

        used = set()


        for i in range(
            len(boxes)
        ):

            if i in used:
                continue


            current_box = boxes[i]


            for j in range(
                i + 1,
                len(boxes)
            ):

                if j in used:
                    continue


                if boxes_should_merge(
                    current_box,
                    boxes[j]
                ):

                    current_box = merge_boxes(
                        current_box,
                        boxes[j]
                    )

                    used.add(j)

                    merged = True


            new_boxes.append(
                current_box
            )

            used.add(i)


        boxes = new_boxes


    # ========================================================
    # CURRENT OBJECTS
    # ========================================================

    current_objects = []


    for x, y, w, h in boxes:

        center_x = x + w // 2
        center_y = y + h // 2


        current_objects.append(
            (
                center_x,
                center_y,
                x,
                y,
                w,
                h
            )
        )


    # ========================================================
    # TRACKING
    # ========================================================

    new_objects = {}
    used_ids = set()


    for (
        center_x,
        center_y,
        x,
        y,
        w,
        h
    ) in current_objects:


        current_center = (
            center_x,
            center_y
        )


        best_id = None
        best_distance = MAX_DISTANCE


        for object_id, data in objects.items():

            if object_id in used_ids:
                continue


            distance = calculate_distance(
                current_center,
                data["center"]
            )


            if distance < best_distance:

                best_distance = distance
                best_id = object_id


        # ====================================================
        # NEW OBJECT
        # ====================================================

        if best_id is None:

            best_id = next_object_id

            next_object_id += 1


            new_objects[best_id] = {

                "center": current_center,

                "box": (
                    x,
                    y,
                    w,
                    h
                ),

                "missing": 0
            }


        # ====================================================
        # EXISTING OBJECT
        # ====================================================

        else:

            previous_x, previous_y, previous_w, previous_h = (
                objects[best_id]["box"]
            )


            smooth_x = int(
                ALPHA * x +
                (1 - ALPHA) * previous_x
            )


            smooth_y = int(
                ALPHA * y +
                (1 - ALPHA) * previous_y
            )


            smooth_w = int(
                ALPHA * w +
                (1 - ALPHA) * previous_w
            )


            smooth_h = int(
                ALPHA * h +
                (1 - ALPHA) * previous_h
            )


            smooth_center = (
                smooth_x + smooth_w // 2,
                smooth_y + smooth_h // 2
            )


            new_objects[best_id] = {

                "center": smooth_center,

                "box": (
                    smooth_x,
                    smooth_y,
                    smooth_w,
                    smooth_h
                ),

                "missing": 0
            }


        used_ids.add(
            best_id
        )


    # ========================================================
    # KEEP TEMPORARILY LOST OBJECTS
    # ========================================================

    for object_id, data in objects.items():

        if object_id in used_ids:
            continue


        missing = (
            data["missing"] + 1
        )


        if missing <= MAX_MISSING:

            new_objects[object_id] = {

                "center": data["center"],

                "box": data["box"],

                "missing": missing
            }


    objects = new_objects


    # ========================================================
    # DRAW DETECTIONS
    # ========================================================

    for object_id, data in objects.items():

        x, y, w, h = data["box"]

        center_x, center_y = data["center"]


        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )


        cv2.circle(
            frame,
            (center_x, center_y),
            5,
            (0, 0, 255),
            -1
        )


        cv2.putText(
            frame,
            f"ID: {object_id}",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )


    # ========================================================
    # INFORMATION
    # ========================================================

    cv2.putText(
        frame,
        f"Frame: {frame_number}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    cv2.putText(
        frame,
        f"Objects: {len(objects)}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    # ========================================================
    # DISPLAY ONLY TWO WINDOWS
    # ========================================================

    cv2.imshow(
        "Mask",
        final_mask
    )

    cv2.imshow(
        "Detection",
        frame
    )


    # ========================================================
    # QUIT
    # ========================================================

    key = cv2.waitKey(1) & 0xFF

    if key == ord("q") or key == 27:
        break


    # ========================================================
    # UPDATE
    # ========================================================

    previous_gray = current_gray.copy()

    frame_number += 1


# ============================================================
# CLEANUP
# ============================================================

cap.release()
cv2.destroyAllWindows()

print("Finished.")
```
