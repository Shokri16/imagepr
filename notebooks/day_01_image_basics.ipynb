```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\shok\Original_lena512.jpg", cv2.IMREAD_GRAYSCALE)

# Histogram equalization
equalized = cv2.equalizeHist(image)

# Histograms
hist = [np.sum(image == i) for i in range(256)]
hist_eq = [np.sum(equalized == i) for i in range(256)]

# Images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap="gray")
plt.title("Equalized")
plt.axis("off")

plt.show()

# Histograms
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(hist)
plt.title("Original Histogram")
plt.xlim(0, 255)

plt.subplot(1, 2, 2)
plt.plot(hist_eq)
plt.title("Equalized Histogram")
plt.xlim(0, 255)

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\shok\lena5.jpg")

# Convert BGR → YCrCb
ycrcb = cv2.cvtColor(image, cv2.COLOR_BGR2YCrCb)

y, cr, cb = cv2.split(ycrcb)

# Equalize brightness channel
y_eq = cv2.equalizeHist(y)

# Combine channels
equalized = cv2.merge([y_eq, cr, cb])

# Convert YCrCb → RGB for matplotlib
original_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
equalized_rgb = cv2.cvtColor(equalized, cv2.COLOR_YCrCb2RGB)

# Show colored images
plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Color")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized_rgb)
plt.title("Equalized Color")
plt.axis("off")

plt.show()
```
