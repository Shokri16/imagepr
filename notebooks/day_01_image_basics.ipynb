```python
import numpy as np
import matplotlib.pyplot as plt

image = np.zeros((400, 400), dtype=np.uint8)

image[75:325, 175:225] = 150
image[175:225, 75:325] = 255

image[10:390, 10:20] = 255
image[10:20, 10:390] = 255 
image[10:390, 380:390] = 255
image[380:390, 10:390] = 255


plt.imshow(image, cmap="magma")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

image = np.zeros((400, 400), dtype=np.uint8)

image[50:350, 30:50] = 255
image[50:350, 330:350] = 255

for i in range(50, 190):
    image[i, i-20:i+20] = 255

for i in range(50, 190):
    image[i, 400 - i - 20:400 - i+20 ] = 255





plt.imshow(image, cmap="magma")
plt.axis("off")
plt.show()
```
