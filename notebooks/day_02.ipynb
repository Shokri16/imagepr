```python
import cv2
import math

# Load image
image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\shok\day 2 shapes (1).jpeg"
)

if image is None:
    print("Error: Could not load the image.")
    exit()

# Get image dimensions
height, width = image.shape[:2]

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Reduce noise
gray = cv2.medianBlur(gray, 5)

# Edge detection
edges = cv2.Canny(gray, 50, 150)

# Detect circles
circles = cv2.HoughCircles(
    gray,
    cv2.HOUGH_GRADIENT,
    dp=1,
    minDist=20,
    param1=100,
    param2=30,
    minRadius=10,
    maxRadius=100
)

count = 0 if circles is None else len(circles[0])

print("Number of circles:", count)

if circles is not None:

    circles = circles[0].astype(int)

    for i, (x, y, r) in enumerate(circles):

        # Calculate diameter and area
        diameter = 2 * r
        area = math.pi * (r ** 2)

        # Draw circle
        cv2.circle(
            image,
            (x, y),
            r,
            (0, 255, 0),
            3
        )

        # Black center dot
        cv2.circle(
            image,
            (x, y),
            5,
            (0, 0, 0),
            -1
        )

        # ------------------------------------------------
        # TEXT POSITION
        # Automatically keep text inside image
        # ------------------------------------------------

        # Default text position
        text_x = x - 65
        text_y = y - r - 35

        # If circle is near the LEFT edge
        if text_x < 10:
            text_x = 10

        # If circle is near the RIGHT edge
        if text_x + 180 > width:
            text_x = width - 190

        # If text is above the image
        if text_y < 25:
            text_y = y + r + 25

        # Make sure text doesn't go below image
        if text_y + 60 > height:
            text_y = height - 65

        # Circle number
        cv2.putText(
            image,
            f"Circle {i + 1}",
            (text_x, text_y),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )

        # Radius and diameter
        cv2.putText(
            image,
            f"R={r}px D={diameter}px",
            (text_x, text_y + 25),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (0, 0, 255),
            2
        )

        # Area
        cv2.putText(
            image,
            f"Area={area:.2f}px^2",
            (text_x, text_y + 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (0, 0, 255),
            2
        )

        # Print information
        print(
            f"Circle {i + 1}: "
            f"Center=({x}, {y}), "
            f"Radius={r}px, "
            f"Diameter={diameter}px, "
            f"Area={area:.2f}px^2"
        )

# Show edge detection
cv2.imshow("Edge Detection", edges)

# Show detected circles
cv2.imshow("Detected Circles", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import math
import numpy as np

# ============================================
# LOAD IMAGE
# ============================================

image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\shok\day 2 shapes (1).jpeg"
)

if image is None:
    print("Error: Could not load the image.")
    exit()

output = image.copy()

height, width = image.shape[:2]

# ============================================
# GRAYSCALE + BLUR
# ============================================

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

gray = cv2.GaussianBlur(
    gray,
    (5, 5),
    0
)

# ============================================
# EDGE DETECTION
# ============================================

edges = cv2.Canny(
    gray,
    50,
    150
)

# Close small gaps
kernel = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (3, 3)
)

edges_closed = cv2.morphologyEx(
    edges,
    cv2.MORPH_CLOSE,
    kernel
)

# ============================================
# FIND CONTOURS
# ============================================

contours, _ = cv2.findContours(
    edges_closed,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

circle_count = 0
triangle_count = 0
rectangle_count = 0

# ============================================
# PROCESS SHAPES
# ============================================

for contour in contours:

    contour_area = cv2.contourArea(contour)

    # Ignore small noise
    if contour_area < 500:
        continue

    perimeter = cv2.arcLength(
        contour,
        True
    )

    # Approximate shape
    epsilon = 0.03 * perimeter

    approx = cv2.approxPolyDP(
        contour,
        epsilon,
        True
    )

    vertices = len(approx)

    x, y, w, h = cv2.boundingRect(contour)

    # ========================================
    # CIRCLE
    # ========================================

    circularity = 0

    if perimeter > 0:
        circularity = (
            4 * math.pi * contour_area
            / (perimeter ** 2)
        )

    if circularity > 0.75 and vertices > 6:

        circle_count += 1

        (cx, cy), radius = cv2.minEnclosingCircle(
            contour
        )

        cx = int(cx)
        cy = int(cy)
        radius = int(radius)

        # Area
        area = math.pi * radius ** 2

        # Circumference
        circumference = 2 * math.pi * radius

        # Draw circle
        cv2.circle(
            output,
            (cx, cy),
            radius,
            (0, 255, 0),
            3
        )

        # Black center
        cv2.circle(
            output,
            (cx, cy),
            5,
            (0, 0, 0),
            -1
        )

        # Small label only
        label = f"Circle {circle_count}"

        label_x = max(10, min(cx - 40, width - 120))
        label_y = max(25, cy - radius - 10)

        cv2.putText(
            output,
            label,
            (label_x, label_y),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )

        # Console information
        print("\n-----------------------------")
        print(f"Circle {circle_count}")
        print(f"Center       : ({cx}, {cy})")
        print(f"Radius       : {radius} px")
        print(f"Area         : {area:.2f} px²")
        print(f"Circumference: {circumference:.2f} px")

    # ========================================
    # TRIANGLE
    # ========================================

    elif vertices == 3:

        triangle_count += 1

        triangle_area = cv2.contourArea(
            contour
        )

        triangle_perimeter = cv2.arcLength(
            contour,
            True
        )

        # Draw triangle
        cv2.drawContours(
            output,
            [approx],
            -1,
            (0, 255, 255),
            3
        )

        # Center
        M = cv2.moments(contour)

        if M["m00"] != 0:
            cx = int(M["m10"] / M["m00"])
            cy = int(M["m01"] / M["m00"])
        else:
            cx = x + w // 2
            cy = y + h // 2

        # Small label
        label = f"Triangle {triangle_count}"

        label_x = max(10, min(cx - 50, width - 150))
        label_y = max(25, cy)

        cv2.putText(
            output,
            label,
            (label_x, label_y),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )

        # Console information
        print("\n-----------------------------")
        print(f"Triangle {triangle_count}")
        print(f"Area     : {triangle_area:.2f} px²")
        print(f"Perimeter: {triangle_perimeter:.2f} px")

    # ========================================
    # RECTANGLE
    # ========================================

    elif vertices == 4:

        # Check if angles are close to 90°
        points = approx.reshape(4, 2)

        angles = []

        for i in range(4):

            p1 = points[i - 1].astype(float)
            p2 = points[i].astype(float)
            p3 = points[(i + 1) % 4].astype(float)

            v1 = p1 - p2
            v2 = p3 - p2

            denominator = (
                np.linalg.norm(v1)
                * np.linalg.norm(v2)
            )

            if denominator == 0:
                continue

            cosine = np.dot(v1, v2) / denominator

            cosine = np.clip(
                cosine,
                -1,
                1
            )

            angle = math.degrees(
                math.acos(cosine)
            )

            angles.append(angle)

        if len(angles) == 4 and all(
            abs(angle - 90) < 20
            for angle in angles
        ):

            rectangle_count += 1

            # Draw rectangle
            cv2.drawContours(
                output,
                [approx],
                -1,
                (255, 0, 255),
                3
            )

            # More accurate rectangle dimensions
            rect = cv2.minAreaRect(contour)

            rect_width, rect_height = rect[1]

            rect_area = rect_width * rect_height

            rect_perimeter = (
                2 * (rect_width + rect_height)
            )

            # Center
            M = cv2.moments(contour)

            if M["m00"] != 0:
                cx = int(M["m10"] / M["m00"])
                cy = int(M["m01"] / M["m00"])
            else:
                cx = x + w // 2
                cy = y + h // 2

            # Small label
            label = f"Rectangle {rectangle_count}"

            label_x = max(
                10,
                min(cx - 55, width - 170)
            )

            label_y = max(
                25,
                cy
            )

            cv2.putText(
                output,
                label,
                (label_x, label_y),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (255, 0, 0),
                2
            )

            # Console information
            print("\n-----------------------------")
            print(f"Rectangle {rectangle_count}")
            print(f"Width    : {rect_width:.2f} px")
            print(f"Height   : {rect_height:.2f} px")
            print(f"Area     : {rect_area:.2f} px²")
            print(f"Perimeter: {rect_perimeter:.2f} px")


# ============================================
# FINAL RESULTS
# ============================================

print("\n=============================")
print("DETECTION SUMMARY")
print("=============================")
print(f"Circles   : {circle_count}")
print(f"Triangles : {triangle_count}")
print(f"Rectangles: {rectangle_count}")
print("=============================")

# ============================================
# DISPLAY
# ============================================

cv2.imshow(
    "Edge Detection",
    edges
)

cv2.imshow(
    "Detected Shapes",
    output
)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\shok\left_half.jpg"
)

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, binary = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY
)

kernel = np.ones((5, 5), np.uint8)

erosion = cv2.erode(binary, kernel, iterations=1)
dilation = cv2.dilate(binary, kernel, iterations=1)
erosion_dilation = cv2.dilate(erosion, kernel, iterations=1)
dilation_erosion = cv2.erode(dilation, kernel, iterations=1)

# Resize images
scale = 0.5

binary = cv2.resize(binary, None, fx=scale, fy=scale)
erosion = cv2.resize(erosion, None, fx=scale, fy=scale)
dilation = cv2.resize(dilation, None, fx=scale, fy=scale)
erosion_dilation = cv2.resize(erosion_dilation, None, fx=scale, fy=scale)
dilation_erosion = cv2.resize(dilation_erosion, None, fx=scale, fy=scale)

cv2.imshow("Original", binary)
cv2.waitKey(0)

cv2.imshow("Erosion", erosion)
cv2.waitKey(0)

cv2.imshow("Dilation", dilation)
cv2.waitKey(0)

cv2.imshow("Erosion then Dilation", erosion_dilation)
cv2.waitKey(0)

cv2.imshow("Dilation then Erosion", dilation_erosion)
cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\shok\lena5.jpg"
)

if image is None:
    print("Error: Could not load image.")
    exit()

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, binary_fixed = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY
)

otsu_value, binary_otsu = cv2.threshold(
    gray, 0, 255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

binary_adaptive_mean = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

binary_adaptive_gaussian = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

triangle_value, binary_triangle = cv2.threshold(
    gray, 0, 255,
    cv2.THRESH_BINARY + cv2.THRESH_TRIANGLE
)

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(binary_fixed, cmap="gray")
plt.title("Fixed (T=127)")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(binary_otsu, cmap="gray")
plt.title(f"Otsu (T={otsu_value:.0f})")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(binary_adaptive_mean, cmap="gray")
plt.title("Adaptive Mean")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(binary_adaptive_gaussian, cmap="gray")
plt.title("Adaptive Gaussian")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(binary_triangle, cmap="gray")
plt.title(f"Triangle (T={triangle_value:.0f})")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np

# =========================
# 1. Read image
# =========================
image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\shok\day 2 (3).jpeg"
)

# Check if image was loaded
if image is None:
    print("ERROR: Image could not be loaded.")
    print("Check the image path.")
    exit()

# =========================
# 2. Resize image
# =========================
scale = 0.5

display_image = cv2.resize(
    image,
    None,
    fx=scale,
    fy=scale
)

# =========================
# 3. Convert BGR -> HSV
# =========================
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# =========================
# 4. OTSU THRESHOLD
# =========================
hue = hsv[:, :, 0]

otsu_value, otsu_mask = cv2.threshold(
    hue,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

otsu_mask_display = cv2.resize(
    otsu_mask,
    None,
    fx=scale,
    fy=scale
)

# =========================
# 5. TRIANGLE THRESHOLD
# =========================
triangle_value, triangle_mask = cv2.threshold(
    hue,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_TRIANGLE
)

triangle_mask_display = cv2.resize(
    triangle_mask,
    None,
    fx=scale,
    fy=scale
)

# =========================
# 6. K-MEANS CLUSTERING
# =========================

# Convert HSV pixels to float32
pixels = hsv.reshape((-1, 3))
pixels = np.float32(pixels)

# Number of clusters
K = 3

criteria = (
    cv2.TERM_CRITERIA_EPS +
    cv2.TERM_CRITERIA_MAX_ITER,
    100,
    0.2
)

_, labels, centers = cv2.kmeans(
    pixels,
    K,
    None,
    criteria,
    10,
    cv2.KMEANS_RANDOM_CENTERS
)

# Convert labels back to image shape
labels = labels.reshape(hsv.shape[:2])

# =========================
# 7. Display K-Means clusters
# =========================

cluster_images = []

for i in range(K):

    mask = np.uint8(labels == i) * 255

    # Use mask to extract the cluster
    result = cv2.bitwise_and(
        image,
        image,
        mask=mask
    )

    result = cv2.resize(
        result,
        None,
        fx=scale,
        fy=scale
    )

    cluster_images.append(result)

# =========================
# 8. Show images
# =========================

cv2.imshow("Original Image", display_image)

cv2.imshow(
    "Otsu Automatic Threshold",
    otsu_mask_display
)

cv2.imshow(
    "Triangle Automatic Threshold",
    triangle_mask_display
)

for i in range(K):
    cv2.imshow(
        f"K-Means Cluster {i + 1}",
        cluster_images[i]
    )

print("Otsu threshold =", otsu_value)
print("Triangle threshold =", triangle_value)
print("K-Means centers:")
print(centers)

print("\nPress any key to close the images.")

cv2.waitKey(0)
cv2.destroyAllWindows()
```
